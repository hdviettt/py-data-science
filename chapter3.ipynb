### Q1. Get a number and verify if it's divisible by 5

```python
def divisible_by_five(x):
    if x % 5 == 0:
        return True
    else: 
        return False

divisible_by_five(int(input()))
```

### Q2. Get a number and verify if it is divisible by 5 or by 3

```python
def q2(x):
    if x % 5 == 0 or x % 3 == 0:
        return True
    else:
        return False

q2(int(input()))
```

### Q3. Get a year and verify whether it is a leap or a common year. A year is a leap year if it is divisble by 4, except those years divisble by 100 are not leap years unless they are also divisible by 400 

```python
def year_indentify(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False

year_indentify(int(input()))
```

### Q4. Get a date and check if it is a magic date. A magic date is when you multiply month by day and get the year. For example, February 12th, 2024 is a magical date: 2 * 12 = 24

```python
''' 
Golden case: 
- Day: 12
- Month: 02
- Year: 2024
'''

def magic_date(date, month, year):
    if date * month == int(str(year)[-2:]):
        return True
    else:
        return False

magic_date(int(input("Enter date")), int(input("Enter month")), int(input("Enter year")))
```
