### Q3. Write a Python program to get a number that contains 3 digits and then display each digit on a single line

```python
q3_input = int(input())

## checking whether the input has 3 digits
def check_three_digits(x):
    if len(str(x)) == 3:
        return True
    else:
        return False

## display each digit on single line
def display_digits(x):
    if check_three_digits(x):
        for digit in str(x):
            print(digit)

display_digits(q3_input)
```

### Q5. Find the roots of a quadratic equation

```python
'''
Quadratic equation: ax^2 + bx + c = 0 (a != 0)
Roots: x = (-b + sqrt(b^2 - 4ac)) / 2a or x = (-b - sqrt(b^2 - 4ac)) / 2a

The program should take the value of a, b, and c as user inputs and then calculate x
'''

q5_a = float(input("enter a:"))
q5_b = float(input("enter b:"))
q5_c = float(input("enter c:"))

def calc_q5_x(a, b, c):
    discriminant = b ** 2 - 4 * a * c
    if discriminant < 0:
        return "No real roots"
    elif discriminant == 0:
        x = -b / (2 * a)
        return x
    else:
        x1 = (-b + math.sqrt(discriminant)) / (2 * a)
        x2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return x1, x2

calc_q5_x(q5_a, q5_b, q5_c)
```
