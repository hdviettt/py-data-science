### Q3. Write a Python program to get a number that contains 3 digits and then display each digit on a single line

```python
q3_input = int(input())

## checking whether the input has 3 digits
def check_three_digits(x):
    if len(str(x)) == 3:
        return True
    else:
        return False

## display each digit on single line
def display_digits(x):
    if check_three_digits(x):
        for digit in str(x):
            print(digit)

display_digits(q3_input)
```

### Q5. Find the roots of a quadratic equation

```python
'''
Quadratic equation: ax^2 + bx + c = 0 (a != 0)
Roots: x = (-b + sqrt(b^2 - 4ac)) / 2a or x = (-b - sqrt(b^2 - 4ac)) / 2a

The program should take the value of a, b, and c as user inputs and then calculate x
'''

q5_a = float(input("enter a:"))
q5_b = float(input("enter b:"))
q5_c = float(input("enter c:"))

def calc_q5_x(a, b, c):
    discriminant = b ** 2 - 4 * a * c
    if discriminant < 0:
        return "No real roots"
    elif discriminant == 0:
        x = -b / (2 * a)
        return x
    else:
        x1 = (-b + math.sqrt(discriminant)) / (2 * a)
        x2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return x1, x2

calc_q5_x(q5_a, q5_b, q5_c)
```

### Q7. get today's data as three integers and display it as m/day/year

```python
'''
Objective:
- Get 3 integers
- Map to date as "m/day/year"
- For example: "3, 15, 24" should be "March/15/2024"

Data guardrails:
- Date should fall between 1 and 31
- Month should be between 1 and 12
- Year can be any positive integer, the program will add 2000 to the year if it's less than 100
- For simplicity purposes, this will not handle cases like February only having 28 or 29 days on leap year
'''


## Handling date
def date():
    date = int(input("Enter date:"))
    if 1 <= date <= 31:
        return date
    else:
        return "Invalid date"

## handling month
def month():
    month = int(input("Enter month:"))
    if 1 <= month <= 12:
        return month
    else:
        return "Invalid month"

## handling year
def year():
    year = int(input("Enter year:"))
    if year < 100:
        return year + 2000
    elif year >= 100:
        return year
    else:
        return "Invalid year"

## Running the program
print(f"{date()}/{month()}/{year()}")
```

### Q10. Generate an even random number between one and one hundred

```python
import random

## Recursion version

def generate_even_number():
    x = random.randint(1, 100)
    if x % 2 == 0:
        return x
    else:
        return generate_even_number()

generate_even_number()
```
